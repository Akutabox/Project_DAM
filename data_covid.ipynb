```python
import pandas as pd

path = '/content/drive/MyDrive/covid19.csv'
corona_dataset = pd.read_csv(path)
```

```python
corona_dataset.shape
```

```python
corona_dataset.head(70)
```

```python
corona_dataset.drop(['Lat', 'Long'], axis=1, inplace=True, errors='ignore')
```

```python
corona_dataset.head(10)
```

```python
corona_dataset_aggregated = corona_dataset.groupby('Country/Region').sum()
corona_dataset_aggregated.head(10)
```

```python
corona_dataset_aggregated.shape
```

```python
corona_dataset_aggregated.loc['Japan'].plot()
corona_dataset_aggregated.loc['India'].plot()
corona_dataset_aggregated.loc['Italy'].plot()
corona_dataset_aggregated.loc['US'].plot()
corona_dataset_aggregated.loc['Spain'].plot()
plt.legend()
```

```python
corona_dataset_aggregated.loc['Spain'].diff().plot()
corona_dataset_aggregated.loc['Italy'].diff().plot()
corona_dataset_aggregated.loc['US'].diff().plot()
corona_dataset_aggregated.loc['India'].diff().plot()
corona_dataset_aggregated.loc['Japan'].diff().plot()
plt.legend()
```
